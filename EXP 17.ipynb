```python
%%writefile input.txt
Cris
Leo
Benzie
Sura
Ney
Bale
```

```python
#Copy odd lines of one file to another.
with open("input.txt","r") as src:
    lines=src.readlines()
with open("odd_lines.txt","w") as dest:
    for i in range(len(lines)):
        if i % 2 ==0:
            dest.write(lines[i])
print("Odd lines copied successfully!")
```

```python
%%writefile studentdetail.csv
Name,Age,Place
```

```python
#Write a python dictionary to a csv file,read it,and display its content.
import csv
studentdetail=[
    {'Name':'Dineesh','Age':20,'Place':'philadhelphia'},
    {'Name':'Sumesh','Age':21,'Place':'oklomo'},
    {'Name':'Suresh','Age':12,'Place':'berlin'}]
with open("studentdetail.csv","w",newline="") as file:
    fields=['Name','Age','Place']
    writer=csv.DictWriter(file,fieldnames=fields)
    writer.writeheader()
    writer.writerows(studentdetail)
print("Dictionary written successfully!")
```
