```python
# 3rd element and 3rd element from last of a tuple
t = tuple(input("Enter tuple elements (comma-separated): ").split(","))
print("3rd element:", t[2])
print("3rd element from last:", t[-3])
```

```python
#accept number from user and generate a tuple and find it's average 
num = input("Enter numbers(comma-seperated): ")
t = tuple(int(x)for x in num.split(","))
average = sum(t) / len(t)
print("Tuple: :", t)
print("Average: ",average)
```
