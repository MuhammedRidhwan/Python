```python
%%writefile Rectangle.py
def Area(l,b):
    return l*b
def Perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile Circle.py
import math

def Area(r):
    return math.pi * r * r

def Perimeter(r):
    return 2 * math.pi * r
```

```python
#main_program
import Rectangle
import Circle
while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")
    ch=int(input("Enter your choice:"))
    if ch==1:
        l=int(input("Enter the length of Rectangle:"))
        b=int(input("Enter the breadth of Rectangle:"))
        print("Area of Rectangle =", Rectangle.Area(l, b))
        print("Perimeter of Rectangle =", Rectangle.Perimeter(l, b))
        print()
    elif ch == 2:
        r = int(input("Enter the radius of Circle: "))
        print("Area of Circle =", round(Circle.Area(r), 2))
        print("Perimeter of Circle =", round(Circle.Perimeter(r), 2))
        print()
    elif ch == 3:
        print("Exiting program...")
        break
```
